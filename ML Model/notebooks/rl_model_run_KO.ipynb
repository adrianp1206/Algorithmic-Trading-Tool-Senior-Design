```python
# main.py
import sys
import os
import time

# Get the absolute path to the `src` directory
src_path = os.path.abspath(os.path.join('..', 'src'))

# Add the `src` directory to the Python path
if src_path not in sys.path:
    sys.path.append(src_path)

import pandas as pd
from trading_env import StockTradingEnv
from dqn_agent import DQNAgent
from train_rl import train_dqn_agent

    # Example data with columns: [Open, High, Low, Close, Volume, XGB_Pred, XGB_Prob_Up]
df = pd.read_csv("KO_RL_input.csv")
print("Read CSV")

    # Create the environment
env = StockTradingEnv(df, initial_balance=10000)
print("Set up env")

    # Create the DQN agent
agent = DQNAgent(state_size=env.state_size, action_size=len(env.action_space))
print("Agent ready")

time.sleep(10)
    # Train the agent
train_dqn_agent(agent, env, episodes=1, save_path="ko_dqn.h5")

    # After training, do a single run to see the final decisions
state = env.reset()
done = False
daily_actions = []
while not done:
    action = agent.act(state)
    next_state, reward, done, _ = env.step(action)
    daily_actions.append((env.current_step, action, reward))
    state = next_state
    
    # Print out the agent's daily actions
action_map = {0: "HOLD", 1: "BUY", 2: "SELL"}
for step, action, reward in daily_actions:
    if step < len(df):
        date_str = df.iloc[step]['Date']
        print(f"Day {date_str}: Action={action_map[action]}, Reward={reward:.2f}")
    else:
        print("End of data.")
```
